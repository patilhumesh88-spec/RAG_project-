```
Loader=PyPDFLoader("/content/20240716890312078.pdf")
Pages=Loader.load()
```

```
Text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
docs = Text_splitter.split_documents(Pages)
```

```
vector_store = Chroma.from_documents(documents=docs, embedding=embeddings , collection_name="IN_const_collection")
```
